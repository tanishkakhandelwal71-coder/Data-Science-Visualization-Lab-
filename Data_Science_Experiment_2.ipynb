```python
# Q.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)
```

```python
# Q.5
def swap_message(val1, val2):
    temp = val1
    val1 = val2
    val2 = temp

    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)


x = 6
y = 3

swap_message(x, y)

print("Outside function:")
print("x =", x)
print("y =", y)
```

8. Scope — Local vs Global Variables

```python
# Q.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)
```

val is a parameter of the function, so it is a local variable. It only exists while my_abs() is running.

```python
# Q.2
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q.3
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
```

because python assumes counter is a local variable inside the function.

But before assigning it, Python tries to read its value.

```python
# Q.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
