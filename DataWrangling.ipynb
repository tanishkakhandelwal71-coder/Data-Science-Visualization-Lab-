```python
# Method 1: combine_first()
result1 = original_indexed.combine_first(updated_indexed)

# Method 2: combine with np.fmax
result2 = original_indexed.combine(
    updated_indexed,
    lambda x, y: np.fmax(x, y)
)

# Side-by-side comparison
q28 = pd.DataFrame({
    'combine_first': result1['Marks'],
    'combine_fmax': result2['Marks']
})

display(q28)
```

```python
different = q28[
    q28['combine_first'] != q28['combine_fmax']
]

display(different)
```
