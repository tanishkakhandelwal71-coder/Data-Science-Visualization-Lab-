```python
def sales_range(x):
    return x.max() - x.min()

df.groupby('product')['units_sold'].agg(sales_range)
```

```python
pivot1 = pd.pivot_table(
    df,
    values='units_sold',
    index='branch',
    columns='product',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='Total'
)

pivot1
```

```python
pivot2 = pd.pivot_table(
    df,
    values=['discount_pct', 'customer_rating'],
    index='branch',
    columns='product',
    aggfunc='mean'
)

pivot2
```

```python
# Display before replacement
print("Before replacement:")
print(df[['branch', 'product', 'customer_rating']])
```

```python
# Replace missing ratings with branch-product average
df['customer_rating'] = df['customer_rating'].fillna(
    df.groupby(['branch', 'product'])['customer_rating'].transform('mean')
)
```

```python
# Display after replacement
print("After replacement:")
print(df[['branch', 'product', 'customer_rating']])
```

```python
def weighted_discount(group):
    return (
        (group['units_sold'] * group['discount_pct']).sum()
        / group['units_sold'].sum()
    )

df.groupby('branch').apply(weighted_discount)
```

```python
top3 = df.groupby('branch', group_keys=False).apply(
    lambda x: x.nlargest(3, 'units_sold')
)

top3[['branch', 'product', 'day', 'units_sold']]
```

```python
branch_total = df.groupby('branch')['units_sold'].transform('sum')

df['sales_contribution_pct'] = (
    df['units_sold'] / branch_total * 100
)
```

```python
df[['branch', 'product', 'units_sold', 'sales_contribution_pct']]
```

```python
df.groupby('branch')['sales_contribution_pct'].sum()
```

```python
df['sales_category'] = pd.qcut(
    df['units_sold'],
    q=4,
    labels=[
        'Low Sales',
        'Below Average Sales',
        'Above Average Sales',
        'High Sales'
    ]
)
```

```python
df.groupby('sales_category', observed=False).agg(
    total_units_sold=('units_sold', 'sum'),
    average_discount=('discount_pct', 'mean'),
    average_customer_rating=('customer_rating', 'mean'),
    average_delivery_time=('delivery_minutes', 'mean'),
    number_of_records=('units_sold', 'count')
)
```
