```python
# Q.29
inventory_df = pd.DataFrame({
    "Item":['pen','rubber','Scale','bag'],
    "Quantity": [2,2,1,1],
    "Price": [10,5,20,150]
})
discount = pd.Series([10, 20], index=[0, 2])
inventory_df["Discount"] = discount
print("After adding Discount column:")
print(inventory_df)
del inventory_df["Discount"]
print("After deleting Discount column:")
print(inventory_df)
```
